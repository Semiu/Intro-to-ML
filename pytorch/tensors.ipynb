Models as computation graphs
These are directed graphs that help in expressing and visualizing mathematical expressions. In DL, it lays out the sequence of calculations needed to compute the output of a neural network.

```python
# Implements a forward pass of a simple logistic regression clasifier

import torch.nn.functional as F

y = torch.tensor([1.0])  # true label - expected to be predicted, i.e. actual
x1 = torch.tensor([1.1]) # input feature
w1 = torch.tensor([2.2]) # weight parameter
b = torch.tensor([0.0])  # bias unit

z = x1 * w1 + b          # net input
a = torch.sigmoid(z)     # activation & output - predicted

loss = F.binary_cross_entropy(a, y) #- difference between predicted and actual
print(loss)
```

```python
import torch.nn.functional as F
from torch.autograd import grad

y = torch.tensor([1.0])
x1 = torch.tensor([1.1])
w1 = torch.tensor([2.2], requires_grad=True) # So that PyTorch internally builds the computation graph
b = torch.tensor([0.0], requires_grad=True)

z = x1 * w1 + b
a = torch.sigmoid(z)

loss = F.binary_cross_entropy(a, y)

grad_L_w1 = grad(loss, w1, retain_graph=True) # To retain the gradient, calling the grad function
grad_L_b = grad(loss, b, retain_graph=True)
```

```python
print(grad_L_w1)
print(grad_L_b)
```

```python
# PyTorch has high level tools for  calculating the gradient
```

```python
loss.backward()

print(w1.grad)
print(b.grad)
```
